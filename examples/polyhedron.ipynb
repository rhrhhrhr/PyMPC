```python
import numpy as np
import matplotlib.pyplot as plt
import tmpc


fig1, ax1 = plt.subplots(1, 1)
ax1.grid(True)
ax1.axis("equal")
ax1.set_title("Translation")

p1 = tmpc.Polyhedron(np.array([[1, 0], [-1, 0], [0, 1], [0, -1]]), np.array([1, 0, 1, 0]))
p2 = tmpc.Polyhedron(np.array([[-3, 1], [-1, -1], [1, 0]]), np.array([3, 1, 0]))
p3 = p1 + np.array([0, 2])

p1.plot(ax1, color="b")
p3.plot(ax1, color="r")
```

```python
p5 = tmpc.Polyhedron(np.array([[1, 1], [-1, 0], [0, -1]]), np.array([2, 0, 0]))

fig2, ax2 = plt.subplots()
ax2.grid(True)
ax2.axis("equal")
ax2.set_title("Map")

p5.plot(ax2, color="g")

set_mat = np.array([[1, 0]])
p6 = p5.map(set_mat)
print(p6)
```
